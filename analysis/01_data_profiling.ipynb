```python
import pandas as pd

print("Python + Pandas funcionando correctamente")
```

```python
from pathlib import Path

DATA_PATH = Path("../data/raw")

ventas = pd.read_csv(DATA_PATH / "Ventas.csv")
devoluciones = pd.read_csv(DATA_PATH / "Devoluciones.csv")
precios = pd.read_csv(DATA_PATH / "Precios.csv")
costos = pd.read_csv(DATA_PATH / "Costos.csv", sep=";")
objetivos = pd.read_csv(DATA_PATH / "Objetivos.csv")

print("Archivos cargados correctamente")
```

```python
print("=== VENTAS ===")
print(ventas.columns.tolist())

print("\n=== DEVOLUCIONES ===")
print(devoluciones.columns.tolist())

print("\n=== PRECIOS ===")
print(precios.columns.tolist())

print("\n=== COSTOS ===")
print(costos.columns.tolist())

print("\n=== OBJETIVOS ===")
print(objetivos.columns.tolist())
```

```python
for nombre, df in tablas.items():
    nulos = df.isna().sum()
    nulos = nulos[nulos > 0]

    print(f"\n=== {nombre} ===")

    if len(nulos) == 0:
        print("Sin valores nulos")
    else:
        print(nulos)
```

```python
print("=== CATEGORÍAS DE PRODUCTO ===")
print(ventas["Categoría_Producto"].value_counts(dropna=False))

print("\n=== SUBCATEGORÍAS DE PRODUCTO ===")
print(ventas["Subcategoría_Producto"].value_counts(dropna=False))

print("\n=== CATEGORÍAS DE VENTA ===")
print(ventas["Categoría_Venta"].value_counts(dropna=False))
```

```python
# ==========================================
# REVISIÓN DE FECHAS
# ==========================================

print("=== FECHAS DE VENTAS ===")

ventas["Fecha"] = pd.to_datetime(ventas["Fecha"], errors="coerce")

print(f"Fecha mínima: {ventas['Fecha'].min()}")
print(f"Fecha máxima: {ventas['Fecha'].max()}")

print("\nRegistros con fecha nula:")
print(ventas["Fecha"].isna().sum())

print("\nCantidad de ventas por año:")
print(ventas["Fecha"].dt.year.value_counts().sort_index())
```

```python
# ==========================================
# REVISIÓN DE DEVOLUCIONES
# ==========================================

print("=== FECHAS DE DEVOLUCIONES ===")

devoluciones["Fecha"] = pd.to_datetime(
    devoluciones["Fecha"], 
    errors="coerce"
)

print(f"Fecha mínima: {devoluciones['Fecha'].min()}")
print(f"Fecha máxima: {devoluciones['Fecha'].max()}")

print("\nRegistros con fecha nula:")
print(devoluciones["Fecha"].isna().sum())

print("\nDevoluciones por año:")
print(devoluciones["Fecha"].dt.year.value_counts().sort_index())
```

```python
# ==========================================
# DIAGNÓSTICO DE FECHAS DE DEVOLUCIONES
# ==========================================

print("=== DEVOLUCIONES FUERA DEL PERÍODO 2024-2025 ===")

devoluciones_fuera_periodo = devoluciones[
    (devoluciones["Fecha"].dt.year < 2024) |
    (devoluciones["Fecha"].dt.year > 2025)
]

print(f"Registros fuera del período: {len(devoluciones_fuera_periodo)}")

print("\nPrimeros registros:")
display(devoluciones_fuera_periodo.head(20))
```

```python
print("=== DEVOLUCIONES CON FECHA NULA ===")

devoluciones_nulas = devoluciones[
    devoluciones["Fecha"].isna()
]

print(f"Registros con fecha nula: {len(devoluciones_nulas)}")

display(devoluciones_nulas.head(20))
```

```python
print("=== CALIDAD DE FECHAS EN DEVOLUCIONES ===")

total_devoluciones = len(devoluciones)

fechas_nulas = devoluciones["Fecha"].isna().sum()

fechas_validas = devoluciones["Fecha"].notna().sum()

print(f"Total de devoluciones: {total_devoluciones:,}")
print(f"Fechas nulas: {fechas_nulas:,}")
print(f"Fechas válidas: {fechas_validas:,}")
print(f"% de fechas nulas: {(fechas_nulas / total_devoluciones) * 100:.2f}%")
```

```python
print("=== DEVOLUCIONES CON FECHA NULA ===")

devoluciones_fecha_nula = devoluciones[
    devoluciones["Fecha"].isna()
]

print(f"Cantidad de registros: {len(devoluciones_fecha_nula):,}")

display(
    devoluciones_fecha_nula.head(20)
)
```

```python
print("=== RANGO Y CALIDAD DE FECHAS EN DEVOLUCIONES ===")

print("Fecha mínima:", devoluciones["Fecha"].min())
print("Fecha máxima:", devoluciones["Fecha"].max())

print("\nCantidad de fechas posteriores a 2025-12-31:")

fechas_futuras = devoluciones[
    devoluciones["Fecha"] > pd.Timestamp("2025-12-31")
]

print(len(fechas_futuras))
```

```python
print("=== DEVOLUCIONES FUERA DEL PERÍODO DEL CHALLENGE ===")

devoluciones_fuera_periodo = devoluciones[
    devoluciones["Fecha"] > pd.Timestamp("2025-12-31")
]

print(f"Cantidad de registros: {len(devoluciones_fuera_periodo):,}")

display(
    devoluciones_fuera_periodo[
        [
            "ID_Devolución",
            "ID_Venta",
            "Fecha",
            "ID_Producto",
            "Cantidad_Devuelta",
            "Motivo",
            "Precio_Unitario"
        ]
    ].sort_values("Fecha").head(20)
)
```

```python
print("=== DEVOLUCIONES CON FECHA NULA Y FECHA DE VENTA DISPONIBLE ===")

devoluciones_fecha_nula = devoluciones[
    devoluciones["Fecha"].isna()
].copy()

devoluciones_fecha_nula["Fecha_Venta"] = devoluciones_fecha_nula["ID_Venta"].map(
    ventas.set_index("ID_Venta")["Fecha"]
)

print(f"Devoluciones con fecha nula: {len(devoluciones_fecha_nula):,}")
print(
    f"Devoluciones con fecha de venta disponible: "
    f"{devoluciones_fecha_nula['Fecha_Venta'].notna().sum():,}"
)
print(
    f"Devoluciones sin fecha de venta disponible: "
    f"{devoluciones_fecha_nula['Fecha_Venta'].isna().sum():,}"
)

display(
    devoluciones_fecha_nula[
        [
            "ID_Devolución",
            "ID_Venta",
            "Fecha",
            "Fecha_Venta",
            "ID_Producto",
            "Cantidad_Devuelta",
            "Motivo"
        ]
    ].head(20)
)
```

```python
print("=== DUPLICADOS DE ID_VENTA EN VENTAS ===")

duplicados_id_venta = ventas[
    ventas["ID_Venta"].duplicated(keep=False)
].sort_values("ID_Venta")

print(f"Registros involucrados en IDs de venta duplicados: {len(duplicados_id_venta):,}")
print(f"IDs de venta duplicados: {duplicados_id_venta['ID_Venta'].nunique():,}")

display(
    duplicados_id_venta[
        [
            "ID_Venta",
            "Fecha",
            "ID_Producto",
            "ID_Cliente",
            "ID_Tienda",
            "Cantidad",
            "Precio_Unitario"
        ]
    ].head(20)
)
```

```python
print("=== VALIDACIÓN DE DUPLICADOS DE ID_VENTA ===")

ventas_duplicadas = ventas[
    ventas["ID_Venta"].duplicated(keep=False)
].sort_values("ID_Venta")

print(f"Registros involucrados: {len(ventas_duplicadas):,}")
print(f"IDs de venta duplicados: {ventas_duplicadas['ID_Venta'].nunique():,}")

print("\n¿Los duplicados tienen la misma fecha de venta?")

conflictos_fecha = (
    ventas_duplicadas
    .groupby("ID_Venta")["Fecha"]
    .nunique()
)

print(
    f"IDs duplicados con fechas diferentes: "
    f"{(conflictos_fecha > 1).sum():,}"
)
```

```python
print("=== DUPLICADOS EXACTOS EN VENTAS ===")

columnas_ventas = ventas.columns.tolist()

duplicados_exactos = ventas[
    ventas.duplicated(subset=columnas_ventas, keep=False)
]

print(f"Registros duplicados exactos: {len(duplicados_exactos):,}")
print(
    f"IDs de venta afectados: "
    f"{duplicados_exactos['ID_Venta'].nunique():,}"
)
```

```python
print("=== DEVOLUCIONES 2026 Y FECHA DE LA VENTA ===")

devoluciones_2026 = devoluciones[
    devoluciones["Fecha"] > pd.Timestamp("2025-12-31")
].copy()

devoluciones_2026["Fecha_Venta"] = devoluciones_2026["ID_Venta"].map(
    ventas.drop_duplicates("ID_Venta").set_index("ID_Venta")["Fecha"]
)

print(f"Devoluciones posteriores a 2025: {len(devoluciones_2026):,}")

print("\nFecha de venta:")
print(devoluciones_2026["Fecha_Venta"].describe())

print("\nDevoluciones 2026 cuya venta pertenece a 2024-2025:")

ventas_periodo = devoluciones_2026[
    devoluciones_2026["Fecha_Venta"] <= pd.Timestamp("2025-12-31")
]

print(f"{len(ventas_periodo):,}")
```

```python
print("=== VALORES NULOS ===")

for nombre, df in {
    "Ventas": ventas,
    "Devoluciones": devoluciones,
    "Precios": precios,
    "Costos": costos,
    "Objetivos": objetivos
}.items():

    nulos = df.isna().sum()
    nulos = nulos[nulos > 0]

    print(f"\n--- {nombre} ---")

    if len(nulos) == 0:
        print("Sin valores nulos.")
    else:
        print(nulos)
```

```python
# ==========================================
# INTEGRIDAD: DEVOLUCIONES VS VENTAS
# ==========================================

ids_ventas = set(ventas["ID_Venta"])

devoluciones_sin_venta = devoluciones[
    ~devoluciones["ID_Venta"].isin(ids_ventas)
]

print("=== DEVOLUCIONES SIN VENTA ASOCIADA ===")
print(f"Total: {len(devoluciones_sin_venta):,}")

display(
    devoluciones_sin_venta[
        ["ID_Devolución", "ID_Venta", "Fecha", "ID_Producto", "Cantidad_Devuelta"]
    ].head(20)
)
```

```python
# ==========================================
# RANGOS Y VALORES ANÓMALOS
# ==========================================

print("=== VENTAS ===")
print("Cantidad mínima:", ventas["Cantidad"].min())
print("Cantidad máxima:", ventas["Cantidad"].max())
print("Precio mínimo:", ventas["Precio_Unitario"].min())
print("Precio máximo:", ventas["Precio_Unitario"].max())

print("\n=== DEVOLUCIONES ===")
print("Cantidad devuelta mínima:", devoluciones["Cantidad_Devuelta"].min())
print("Cantidad devuelta máxima:", devoluciones["Cantidad_Devuelta"].max())
print("Precio mínimo:", devoluciones["Precio_Unitario"].min())
print("Precio máximo:", devoluciones["Precio_Unitario"].max())

print("\n=== COSTOS ===")
costo_num = pd.to_numeric(
    costos["Costo_Unitario"].astype(str).str.replace(",", ".", regex=False),
    errors="coerce"
)

print("Costo mínimo:", costo_num.min())
print("Costo máximo:", costo_num.max())

print("\n=== CEROS Y NEGATIVOS ===")
print("Ventas - cantidades <= 0:", (ventas["Cantidad"] <= 0).sum())
print("Ventas - precios <= 0:", (ventas["Precio_Unitario"] <= 0).sum())
print("Devoluciones - cantidades <= 0:", (devoluciones["Cantidad_Devuelta"] <= 0).sum())
print("Devoluciones - precios <= 0:", (devoluciones["Precio_Unitario"] <= 0).sum())
print("Costos - costos <= 0:", (costo_num <= 0).sum())
```
